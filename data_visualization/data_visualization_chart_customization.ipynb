```python
import pandas as pd
```

```python
df = pd.DataFrame({
    "Months": ["Jan", "Feb", "Mar", "Apr", "May", "Jun"],
    "Sales": [1200,1350,1600,1500,1700,1950],
    "Profit": [300,340,400,380,420,500],
    "Visitors": [4000,4300,5000,4800,5200,6000]
})
```

```python
df
```

# Line Styles

- dashed
- solid
- dashdot
- dotted

```python
import matplotlib.pyplot as plt
```

```python
plt.plot(df["Months"], df["Sales"], linestyle = "solid")
plt.show()
```

# Colors

- Color name
- Hex
- RGB

```python
plt.plot(df["Months"], df["Sales"], color = "Red")
plt.show()
```

# Markers

- o
- s
- ^
- x

```python
plt.plot(df["Months"], df["Sales"], marker = "x")
plt.show()
```

# Figure Size and Labels

```python
plt.figure(figsize=(10,5))  #Figure Size
plt.plot(df["Months"], df["Sales"], marker = "x")

#Labels
plt.title("Example", fontsize = 30)
plt.xlabel("Month", fontsize = 20)
plt.ylabel("Sales", fontsize = 20)  # Fontsize

plt.show()
```

# Axis Limits

```python
plt.plot(df["Months"], df["Sales"])

# Limits
plt.ylim(1000, 2000)  
plt.xlim(-1, 6)
plt.show()
```

# Gridlines

```python
plt.plot(df["Months"], df["Sales"])
plt.grid(True, linestyle = "dotted", color = "black", alpha = 0.4)  #Gridline
plt.show()
```

# Legends

```python
plt.plot(df["Months"], df["Sales"], label = "Sales")
plt.plot(df["Months"], df["Profit"], label = "Profit")
plt.legend()
plt.show()
```

# Multiple Subplots

```python
fig, ax = plt.subplots(1,2, figsize=(12,4))

ax[0].plot(df["Months"], df["Sales"])
ax[1].bar(df["Months"], df["Profit"])

plt.show()
```

```python
import plotly.express as px
import plotly.graph_objects as go
```

# Color

```python
fig = px.line(df, x = "Months", y = "Sales", color_discrete_sequence = ["black"])
fig.update_traces(line = dict(dash = "dot"))  # ("longdash")  --  linestyle
fig.update_traces(mode = "lines+markers")    # marker
fig.update_layout(
    plot_bgcolor = "Red",
    paper_bgcolor = "yellow",
    title = "Sales Chart",
    xaxis_title = "Month",              # layout edit
    yaxis_title = "Sales",
    width = 900,
    height = 500
)
fig.update_xaxes(showgrid = True, gridcolor = "black")          # Gridline for |
fig.update_yaxes(showgrid = True, gridcolor = "white")          # Gridline for --
fig.show()
```
