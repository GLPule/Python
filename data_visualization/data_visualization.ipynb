# Grouped bar chart

```python
products = ["A", "B", "C"]
sales_2023 = [400, 500, 450]
sales_2024 = [480, 530, 490]
```

```python
import numpy as np
```

```python
x = np.arange(len(products))
width = 0.35
```

```python
x
```

```python
plt.bar(x - width, sales_2023, width, label= "2023")
plt.bar(x, sales_2024, width, label= "2024")

plt.xticks(x, products)
plt.title("Grouped Sales Comparison")
plt.xlabel("Products")
plt.ylabel("Sales")
plt.legend()
plt.show()
```

# Stacked bar chart

```python
categories = ["A", "B", "C"]
online = [200, 250, 300]
offline = [150, 180, 200]
```

```python
plt.bar(categories, online, label= "Online")
plt.bar(categories, offline, bottom=online, label= "Offline")

plt.title("Sales by Channel")
plt.legend()
plt.show()
```

# Radial Plot

```python
labels = ["A", "B", "C", "D"]
values = [4, 7, 1, 5]
```

```python
angles = np.linspace(0, 2*np.pi, len(values), endpoint= False)
values = np.concatenate((values, [values[0]]))
angles = np.concatenate((angles, [angles[0]]))
```

```python
plt.polar(angles, values)
plt.title("Radial Plot")
plt.show()
```

# Radar Chart

```python
labels = ["Speed", "Power", "Control", "Stability"]
values = [7,6,8,5]

angles = np.linspace(0, 2*np.pi, len(values), endpoint= False)
values = np.concatenate((values, [values[0]]))
angles = np.concatenate((angles, [angles[0]]))
```

```python
plt.polar(angles, values)
plt.fill(angles, values, alpha = 0.2)
plt.title("Radar Chart Example")
plt.show()
```

# Choropleth Map

Regions colored by value, such as population or sales

```python
import plotly.express as px
```

```python
px.data.gapminder()
```

```python
df = px.data.gapminder().query("year == 2007")
```

```python
df
```

```python
fig = px.choropleth(
    df, 
    locations = "iso_alpha",
    color = "lifeExp",
    hover_name="country"
)

fig.show()
```

# Scatter Map

Points plotted on a map using latitude and longitude

```python
df = px.data.carshare()
```

```python
df
```

```python
fig = px.scatter_mapbox(
    df,
    lat = "centroid_lat",
    lon = "centroid_lon",
    size = "car_hours",
    zoom = 10
)

fig.update_layout(mapbox_style = "open-street-map")
fig.show()
```

# Hexbin Map

Geographic bins shaped like hexagons

```python
df = px.data.carshare()
```

```python
df
```

```python
fig = px.density_mapbox(
    df,
    lat = "centroid_lat",
    lon = "centroid_lon",
    radius = 20,
    zoom = 10
)

fig.update_layout(mapbox_style = "open-street-map")
fig.show()
```
