### conditional filtering

- & AND
- | OR
- ~ NOT

```python
# Price greater than 30 AND product tea

df[(df["price"] > 30) & (df["product"] == "tea")]
```

```python
# Price less 40 OR quantity greater than 3

df[(df["price"] < 40) | (df["quantity"] > 3)]
```

```python
# NOT tea

df[~(df["product"] == "tea")]
```

### Pivot Tables

```python
df
```

```python
df.pivot_table(
    values = "Sales",
    index = "Product",
    columns = "Store",
    aggfunc = "sum",
    margins = True
)
```


- **values** -> what you what to summarize
- **index** -> this is the rows
- **columns** -> new columns created
- **aggfunc** -> calculation to apply inside each cell
- **margins** -> to show the total for rows and columns


```python
df
```

```python
df.pivot_table(
    values = "Sales",
    index = "Product",
    columns = "Store",
    aggfunc = ["sum", "mean", "count"]
)
```

#### binning (discretization)

```python
# 0 - 17 child, 18 - 64 adult 65+ senior
# poor, fair, good, excellent

ages = [15,22,37,45,72,16]
df = pd.DataFrame({"age": ages})
df
```

```python
bins = [0, 18, 65, 120]
labels = ["child", "adult", "senior"]
```

```python
df["group"] = pd.cut(df["age"], bins = bins, labels = labels)
```

```python
df
```
